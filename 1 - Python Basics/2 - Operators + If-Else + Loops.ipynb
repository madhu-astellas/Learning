### Operators in Python

* Arithmetic Operators
* Relational Operators
* Logical Operators
* Bitwise Operators
* Assignment Operators
* Membership Operators

```python
# Arithmetic Operators

print(5 + 6)

print(5 - 6)

print(5 * 6)

print(5 / 2)

print(5 // 2)

print(5 % 2)

print(5 ** 2)
```

```python
# Relational Operators

print(4 > 5)

print(4 < 5)

print(4 >= 5)

print(4 <= 5)

print(4 == 5)

print(4 != 5)
```

```python
# Logical Operators

print(1 and 0)

print(1 or 0)

print(not 0)
```

```python
# Bitwise Operators

# bitwise and
print(2 & 3)  # 010 & 011

# bitwise or
print(2 | 3)  # 010 | 011

# bitwise xor
print(2 ^ 3)  # 010 ^ 011 -> Bitwise XOR is a binary operator (^) that compares two numbers bit‑by‑bit and outputs 1 where the bits differ and 0 where they are the same

# bitwise not
print(~3)  # 011 -> 100 and negative

# right shift
print(4 >> 2) # 100 -> 001

# left shift
print(4 << 2) # 100 -> 10000
```

```python
# Assignment Operators

# =
a = 2

a += 2

print(a)
```

```python
a %= 2
print(a)
```

```python
a++ # wont work in python
```

```python
# Membership Operators

# in / not in

print('D' in 'Delhi')

print('d' in 'Delhi')

print('d' not in 'Delhi')

print(1 in [2,3,4,5])
```

### Loops in Python

* Need for Loops

```python
# while loop
number = int(input('Enter a number'))

i = 1
while i<11:    # pythontutor.com
    print(number, '*', i, '=', number * i)
    i += 1 
```

```python
# while loop with else

x = 1

while x < 3:
    print(x)
    x += 1

else:
    print('Limit crossed')
```

```python
# Guessing game
# Generate a random integer between 1 and 100

import random

jackpot = random.randint(1, 100)

guess = int(input('Guess a number'))

if guess < jackpot:
    print("Wrong! Guess higher")
elif guess > jackpot:
    print("Wrong! Guess lower")
else:
    print("Correct")
```

```python
import random

jackpot = random.randint(1, 100)

guess = int(input('Guess a number'))
counter = 1

while guess != jackpot:
    if guess < jackpot:
        print("Wrong! Guess higher")
    else:
        print("Wrong! Guess lower")
    
    guess = int(input('Guess a number'))
    counter += 1
    
else:
    print("Correct")
    print("attempts", counter)
```

#### Program - The current population of a town is 10000. The population of the town is increasing at the rate of 10% per year. You have to write a program to find out the population at the end of each of the last 10 years

```python
curr_pop = 10000

for i in range(10,0,-1):
    print(i, curr_pop)
    curr_pop = curr_pop/1.1
```

#### Sequence sum

#### 1/1! + 2/2! + 3/3! + ...

```python
n = int(input('Enter a number'))

result = 0
fact = 1

for i in range(1, n+1):
    fact = fact * i
    result = result + i/fact

print(result)
```

### Pattern 1
##### *
##### **
##### ***

```python
rows = int(input('Enter number of rows'))

for i in range(1, rows+1):
    for j in range(1, i+1):
        print('*', end="")
    print()
```

##### Pattern 2
###### 1
###### 121
###### 12321
###### 1234321

```python
rows = int(input('Enter number of rows'))

for i in range(1, rows+1):
    for j in range(1, i+1):
        print(j, end="")
    for k in range(i-1, 0, -1):
        print(k, end='')
    print()
```

### Loop Control Statement

* Break
* Continue
* Pass

```python
# break

for i in range(1, 10):
    if i == 5:
        break
    print(i)
```

```python
lower = int(input('Enter lower range'))
upper = int(input('Enter upper range'))

for i in range(lower, upper+1):
    for j in range(2, i):
        if i%j == 0:
            break
    else:
        print(i)
```

```python
# continue
for i in range(1,10):
    if i == 5:
        continue
    print(i)
```

```python
# pass
for i in range(1,10):
    pass
```
