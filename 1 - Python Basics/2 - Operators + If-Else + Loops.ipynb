### Operators in Python

* Arithmetic Operators
* Relational Operators
* Logical Operators
* Bitwise Operators
* Assignment Operators
* Membership Operators

```python
# Arithmetic operators: + - * / // (floor) % (modulo) ** (power).
# Arithmetic Operators
# Output: 11, -1, 30, 2.5, 2, 1, 25 -> note that '/' gives a float (5/2=2.5), '//' the floor (2) and '%' the remainder (1).

print(5 + 6)

print(5 - 6)

print(5 * 6)

print(5 / 2)

print(5 // 2)

print(5 % 2)

print(5 ** 2)
```

```python
# Relational (comparison) operators always return a boolean.
# Relational Operators
# Output: False, True, False, True, False, True -> each comparison shows whether the relation holds.

print(4 > 5)

print(4 < 5)

print(4 >= 5)

print(4 <= 5)

print(4 == 5)

print(4 != 5)
```

```python
# Logical operators: and, or, not.
# Logical Operators
# Output: 0, 1, True -> 'and'/'or' return one of their operands (1 and 0 here) while 'not' returns a bool.

print(1 and 0)

print(1 or 0)

print(not 0)
```

```python
# Bitwise operators act on the binary (bit) representation of integers.
# Bitwise Operators
# Output: 2, 3, 1, -4, 1, 16 -> the outcomes of &, |, ^, ~, >> and << on the binary forms (e.g. ~3 = -4).

# bitwise and
print(2 & 3)  # 010 & 011

# bitwise or
print(2 | 3)  # 010 | 011

# bitwise xor
print(2 ^ 3)  # 010 ^ 011 -> Bitwise XOR is a binary operator (^) that compares two numbers bit‑by‑bit and outputs 1 where the bits differ and 0 where they are the same

# bitwise not
print(~3)  # 011 -> 100 and negative

# right shift
print(4 >> 2) # 100 -> 001

# left shift
print(4 << 2) # 100 -> 10000
```

```python
# Assignment: '=' stores a value; '+=' adds then stores back.
# Assignment Operators
# Output: 4 -> a starts at 2 and '+=' adds 2 again.

# =
a = 2

a += 2

print(a)
```

```python
# '%=' takes the remainder and stores it back into a.
# Output: 0 -> 4 % 2 leaves remainder 0, so a becomes 0.
a %= 2
print(a)
```

```python
# Python has no ++ operator (unlike C/Java) -> SyntaxError.
# Output: SyntaxError - Python has no ++ operator; use a += 1 instead.
a++ # wont work in python
```

```python
# Membership tests: 'in' and 'not in' check for presence (case sensitive).
# Membership Operators
# Output: True, False, True, False -> 'in' is case sensitive ('D' is present, 'd' is not) and 1 is not in the list.

# in / not in

print('D' in 'Delhi')

print('d' in 'Delhi')

print('d' not in 'Delhi')

print(1 in [2,3,4,5])
```

```python
# Program: add the digits of a 3-digit number using % and //.
# Program - Find the sum of a 3 digit number entered by the user
# Output: 12 -> for 345 the digits are pulled off with %10 and //10 and 3+4+5 = 12.

number = int(input('Enter a 3 digit number'))

# 345%10 -> 5
a = number % 10

# 345//10 -> 34
number = number // 10

b = number % 10

number = number // 10

c = number % 10

print(a+b+c)
```

### If-else in Python

```python
# Login check using if/else; indentation defines the block.
# login program and indentation
# email -> madhu.kumar245@gmail.com
# password -> 1234
# Output: Welcome -> the typed email and password both match, so the if-branch runs.

email = input('Enter Email')
password = input('Enter Password')

if email == 'madhu.kumar245@gmail.com' and password == '1234':
    print('Welcome')
else:
    print('Not correct')
```

```python
# elif adds a second condition to test when the first is False.
# Output: Incorrect password -> the email matches but the password does not, so the elif branch runs.
email = input('Enter Email')
password = input('Enter Password')

if email == 'madhu.kumar245@gmail.com' and password == '1234':
    print('Welcome')
elif email == 'madhu.kumar245@gmail.com' and password != '1234':
    print('Incorrect password')
else:
    print('Not correct')
```

```python
# Nested if/else to allow one retry before locking the account.
# Output: Incorrect password then Your account is locked -> the retyped password is also wrong, so the nested else runs.
email = input('Enter Email')
password = input('Enter Password')

if email == 'madhu.kumar245@gmail.com' and password == '1234':
    print('Welcome')
elif email == 'madhu.kumar245@gmail.com' and password != '1234':
    print('Incorrect password')
    password = input('Enter Password again')
    if password == '1234':
        print('Welcome, Finally!')
    else:
        print('Your account is locked')
else:
    print('Not correct')
```

```python
# Program: find the smallest of three numbers.
# 1. Find the min of 3 given numbers
# Output: Smallest is 2 -> a is less than both b and c, so the first branch prints it.

a = int(input('First num'))
b = int(input('Second num'))
c = int(input('Third num'))

if a<b and a<c:
    print('Smallest is', a)
elif b<c:
    print('Smallest is', b)
else:
    print('Smallest is', c)
```

```python
# Program: menu-driven calculator with if/elif/else.
# Menu Driven Calculator
# Output: 110 -> the program applies the selected operator to the two typed numbers.

fnum = int(input('Enter the 1st number'))
snum = int(input('Enter the 2nd number'))

op = input('Enter the operation')

if op == '+':
    print(fnum + snum)
elif op == '-':
    print(fnum - snum)
elif op == '*':
    print(fnum * snum)
else:
    print(fnum / snum)
```

```python
# Program: menu-driven ATM; a triple-quoted string holds the menu text.
# Menu Driven ATM
# Output: balance -> menu choice '2' selects the balance branch.

menu = input("""
Hi! How can I help you?
1. Enter 1 for pin change
2. Enter 2 for balance check
3. Enter 3 for withdrawal
4. Enter 4 for exit
""")

if menu == '1':
    print('pin change')
elif menu == '2':
    print('balance')
else:
    print('exit')
```

### Loops in Python

* Need for Loops
* While Loop

```python
# while loop repeats its body while the condition stays True (multiplication table).
# while loop
# Output: the times table 9x1 ... 9x10 -> the loop repeats while i<11, printing number*i each pass.
number = int(input('Enter a number'))

i = 1
while i<11:    # pythontutor.com
    print(number, '*', i, '=', number * i)
    i += 1 
```

```python
# A while loop may have an else block that runs when the loop ends normally.
# while loop with else
# Output: 1, 2, then Limit crossed -> the loop stops when x<3 becomes False and the else block runs.

x = 1

while x < 3:
    print(x)
    x += 1

else:
    print('Limit crossed')
```

```python
# Program: one-attempt number guessing game.
# Guessing game
# Generate a random integer between 1 and 100
# Output: Wrong! Guess lower -> the guess was greater than the random jackpot.

import random

jackpot = random.randint(1, 100)

guess = int(input('Guess a number'))

if guess < jackpot:
    print("Wrong! Guess higher")
elif guess > jackpot:
    print("Wrong! Guess lower")
else:
    print("Correct")
```

```python
# Program: keep guessing until correct; while/else counts the attempts.
# Output: several hints, then Correct and attempts 6 -> the loop repeats until the guess equals jackpot, counting every try.
import random

jackpot = random.randint(1, 100)

guess = int(input('Guess a number'))
counter = 1

while guess != jackpot:
    if guess < jackpot:
        print("Wrong! Guess higher")
    else:
        print("Wrong! Guess lower")
    
    guess = int(input('Guess a number'))
    counter += 1
    
else:
    print("Correct")
    print("attempts", counter)
```

```python
# for loop over range(start, stop); stop is exclusive.
# For Loop
# Output: 1 to 10 -> range(1,11) stops before 11 (the stop value is excluded).

for i in range(1,11):
    print(i)
```

```python
# The third argument of range() is the step (every 2nd number).
# Output: 1, 3, 5, 7, 9 -> a step of 2 skips every other number.
for i in range(1,11,2):
    print(i)
```

```python
# A step of 4 visits 1, 5 and 9.
# Output: 1, 5, 9 -> a step of 4.
for i in range(1,11,4):
    print(i)
```

```python
# A negative step counts downward.
# Output: 10 down to 1 -> the negative step counts backwards; 0 is excluded.
for i in range(10,0,-1):
    print(i)
```

```python
# for can iterate over any iterable, here a tuple.
# Output: 1, 2, 3, 4, 99 -> the loop visits each item of the tuple in order.
for i in (1,2,3,4,99):
    print(i)
```

#### Program - The current population of a town is 10000. The population of the town is increasing at the rate of 10% per year. You have to write a program to find out the population at the end of each of the last 10 years

```python
# Loop that shrinks a population by 10% each year.
# Output: the counter and the shrinking population -> curr_pop is divided by 1.1 each pass (a 10% drop).
curr_pop = 10000

for i in range(10,0,-1):
    print(i, curr_pop)
    curr_pop = curr_pop/1.1
```

#### Sequence sum

#### 1/1! + 2/2! + 3/3! + ...

```python
# Program: evaluate the series 1/1! + 2/2! + ... + n/n!.
# Output: 2.708333333333333 -> with n=5 the running sum 1/1! + 2/2! + ... + 5/5! freezes at this value.
n = int(input('Enter a number'))

result = 0
fact = 1

for i in range(1, n+1):
    fact = fact * i
    result = result + i/fact

print(result)
```

### Nested Loops

```python
# Nested loops: the inner loop completes for every outer iteration.
# Output: the pairs (1,1) up to (4,4) -> the inner loop finishes for every value of the outer loop, giving 4x4 = 16 lines.
for i in range(1,5):
    for j in range(1,5):
        print(i,j)
```

### Pattern 1
##### *
##### **
##### ***

```python
# Pattern: right triangle of stars using nested loops.
# Output: an increasing triangle of stars -> row i prints i stars; end='' keeps them on one line and print() ends the row.
rows = int(input('Enter number of rows'))

for i in range(1, rows+1):
    for j in range(1, i+1):
        print('*', end="")
    print()
```

##### Pattern 2
###### 1
###### 121
###### 12321
###### 1234321

```python
# Pattern: number pyramid built with two inner loops.
# Output: 1, 121, 12321, 1234321, 123454321 -> each row counts up then back down.
rows = int(input('Enter number of rows'))

for i in range(1, rows+1):
    for j in range(1, i+1):
        print(j, end="")
    for k in range(i-1, 0, -1):
        print(k, end='')
    print()
```

### Loop Control Statement

* Break
* Continue
* Pass

```python
# break exits the loop immediately.
# break
# Output: 1, 2, 3, 4 -> break leaves the loop as soon as i equals 5, so 5..9 are never printed.

for i in range(1, 10):
    if i == 5:
        break
    print(i)
```

```python
# Program: print prime numbers in a range using for/else and break.
# Output: the prime numbers in the range -> the inner loop tries every divisor and for/else prints i only when none divided it (so 1 is also printed).
lower = int(input('Enter lower range'))
upper = int(input('Enter upper range'))

for i in range(lower, upper+1):
    for j in range(2, i):
        if i%j == 0:
            break
    else:
        print(i)
```

```python
# continue skips the remaining statements of the current iteration.
# continue
# Output: 1 to 9 without 5 -> continue skips the rest of the body only when i is 5.
for i in range(1,10):
    if i == 5:
        continue
    print(i)
```

```python
# pass is a no-op placeholder that keeps a block syntactically valid.
# pass
# Output: nothing -> 'pass' does nothing, so the loop runs but prints no values.
for i in range(1,10):
    pass
```
