##### Creating Strings

```python
# Strings can be created with single quotes.
s = 'hello'
print(s)
```

```python
# Double quotes work the same way.
s = "hello"
print(s)
```

```python
# Triple-quoted strings can span multiple lines.
# for multiline strings
s = """hello"""
print(s)
```

```python
# An apostrophe inside single quotes ends the string early -> SyntaxError.
'it's raining outside'
```

```python
# Double quotes avoid the need to escape the apostrophe.
"it's raining outside"
```

```python
# The str() constructor also builds a string.
s = str('hello')
print(s)
```

##### Accessing Substrings from a String

```python
# Indexing starts at 0, so s[0] is the first character.
# Indexing

s = 'hello world'
print(s[0])
```

```python
# s[4] is the fifth character.
print(s[4])
```

```python
# An index outside the string raises an IndexError.
print(s[41])
```

```python
# A negative index counts from the end; -1 is the last character.
# Negative Indexing
print(s[-1])
```

```python
# -2 is the second character from the end.
print(s[-2])
```

```python
# Slicing s[start:stop]; the stop index is excluded.
# Slicing

print(s[0:5])
```

```python
# Characters from index 6 up to (not including) index 11.
print(s[6:11])
```

```python
# Stops before index 10, so the final d is left out.
print(s[6:10])
```

```python
# Leaving out stop slices through to the end.
print(s[2:])
```

```python
# From index 6 to the end.
print(s[6:])
```

```python
# A copy of the whole string.
print(s[:])
```

```python
# The third slice value is the step (every 2nd character).
print(s[0:5:2])
```

```python
# A step of 3.
print(s[0:8:3])
```

```python
# A step of 1 is the default.
print(s[0:5:1])
```

```python
# A negative step with start before stop returns an empty string.
print(s[0:5:-1])
```

```python
# A negative step walks backwards; stop (0) is excluded.
print(s[5:0:-1])
```

```python
# Backwards with a step of 2.
print(s[5:0:-2])
```

```python
# Idiom to reverse a string.
print(s[::-1])
```

```python
# Slice using negative start and stop.
print(s[-5:-1])
```

```python
# The last five characters.
print(s[-5:])
```

```python
# The last five characters, reversed.
print(s[-1:-6:-1])
```

##### Editing and Deleting in Strings

```python
# Set up s again.
s = 'hello world'
```

```python
# Strings are immutable, so item assignment raises a TypeError.
s[0] = 'H'  # Python strings are immutable
```

```python
# Re-create s after the deletion example.
s = 'hello world'
```

```python
# del removes the name; using it afterwards raises a NameError.
del s
print(s)
```

```python
# Slices cannot be deleted -> TypeError.
s = 'hello world'
del s[-1:-5:2]
print(s)
```

##### Operations on Strings
* Arithmetic Operations
* Relational Operations
* Logical Operations
* Loops on Strings
* Membership Operations

```python
# The + operator concatenates strings.
print('delhi' + 'mumbai')
```

```python
# Concatenate with a space in between.
print('delhi' + ' ' + 'mumbai')
```

```python
# The * operator repeats a string.
print('delhi'*5)
```

```python
# Repeating a character to draw a separator line.
print('*'*50)
```

```python
# Equality is case sensitive.
'delhi' == 'mumbai'
```

```python
# Not-equal comparison.
'delhi' != 'delhi'
```

```python
# A difference in case makes the strings unequal.
'delhi' == 'Delhi'
```

```python
# Strings compare lexicographically, character by character.
'mumbai' > 'pune'   # lexiographically comparison
```

```python
# Uppercase letters have smaller codes than lowercase, so P < p.
'Pune' > 'pune'  # https://www.littlebeartools.com/asciicode/
```

```python
# 'and' returns the first falsy operand, otherwise the last operand.
'hello' and 'world'
```

```python
# 'or' returns the first truthy operand.
'hello' or 'world'
```

```python
# An empty string is falsy.
'' and 'world'
```

```python
# So 'or' returns the non-empty string.
'' or 'world'
```

```python
# 'not' negates the truthiness of a value.
not 'world'
```

```python
# An empty string is falsy, so not '' is True.
not ''
```

```python
# A string can be iterated character by character.
for i in 'hello':
    print(i)
```

```python
# The body runs once for each character in the string.
for i in 'delhi':
    print('pune')
```

```python
# Membership is case sensitive.
'D' in 'delhi'
```

```python
# The empty string is considered to be in every string.
'' in 'delhi'
```

```python
# There is no space in the string, so this is False.
' ' in 'delhi'
```

##### Common Functions
* len
* max
* min
* sorted

```python
# len() gives the number of characters.
len('Madhu Dadi')
```

```python
# max() gives the character with the highest code.
max('Madhu Dadi')
```

```python
# min() gives the character with the lowest code.
min('Madhu Dadi')
```

```python
# sorted() returns the characters as a list, in ascending order.
sorted('Madhu Dadi')
```

```python
# reverse=True sorts in descending order.
sorted('Madhu Dadi', reverse=True)
```

##### Capitalize / Title / Upper / Lower / Swapcase

```python
# capitalize() returns a copy with the first character upper-cased.
s = 'hello world'
s.capitalize()
```

```python
# Strings are immutable, so the original s is unchanged.
print(s)
```

```python
# title() upper-cases the first letter of each word.
s.title()
```

```python
# upper() returns an all-uppercase copy.
s.upper()
```

```python
# lower() returns an all-lowercase copy.
s.lower()
```

```python
# swapcase() inverts the case of each character.
s.swapcase()
```

```python
# swapcase() applied to a string literal.
'Madhu Dadi'.swapcase()
```

##### Count / Find / Index

```python
# count() returns how many times a substring occurs.
'my name is madhu'.count('a')
```

```python
# find() returns the index of the first occurrence, or -1 if absent.
'my name is madhu'.find('is')
```

```python
# Not found, so find() returns -1.
'my name is madhu'.find('x')
```

```python
# index() is like find(), but raises ValueError when the substring is absent.
'my name is madhu'.index('is')
```

```python
# Absent substring -> ValueError.
'my name is madhu'.index('x')
```

##### endswith / startswith

```python
# endswith() tests the suffix.
'my name is madhu'.endswith('hu')
```

```python
# False, because the string does not end with a.
'my name is madhu'.endswith('a')
```

```python
# startswith() tests the prefix.
'my name is madhu'.startswith('hu')
```

```python
# True, because the string starts with my.
'my name is madhu'.startswith('my')
```

##### Format

```python
# str.format() fills the {} placeholders with the arguments in order.
name = 'madhu'
gender = 'male'

'Hi, my name is {} and I am a {}'.format(name, gender)
```

```python
# Placeholders are filled in the order the arguments are given.
'Hi, my name is {} and I am a {}'.format(gender, name)
```

```python
# Numbered placeholders refer to arguments by position.
'Hi, my name is {0} and I am a {1}'.format(name, gender)
```

```python
# The numbering can reorder the arguments.
'Hi, my name is {1} and I am a {0}'.format(gender, name)
```

```python
# The same index may be reused in the template.
'Hi, my name is {1} and I am a {1}'.format(gender, name)
```

##### isalnum / isalpha / isdigit / isidentifier

```python
# isalnum() is True when all characters are letters or digits.
'madhu1234'.isalnum()
```

```python
# isdigit() is False because the string also has letters.
'123abc'.isdigit()
```

```python
# A single digit -> True.
'1'.isdigit()
```

```python
# A letter -> False.
'a'.isdigit()
```

```python
# isidentifier() checks whether the text is a valid Python name.
'name'.isidentifier()
```

```python
# An identifier may start with an underscore.
'_name'.isidentifier()
```

```python
# A hyphen is not allowed in an identifier.
'first-name'.isidentifier()
```

```python
# Keywords are still valid identifiers, though best avoided.
'print'.isidentifier()
```

##### Split / Join

```python
# split() with no argument splits on whitespace.
'Hi, my name is madhu'.split()
```

```python
# split(sep) splits on the given separator.
'Hi, my name is madhu'.split(',')
```

```python
# Splitting on 'a' removes the separator and returns the pieces.
'Hi, my name is madhu'.split('a')
```

```python
# join() glues an iterable of strings using the calling string as separator.
" ".join(['Hi,', 'my', 'name', 'is', 'madhu'])
```

```python
# Join with a hyphen separator.
"-".join(['Hi,', 'my', 'name', 'is', 'madhu'])
```

##### Replace

```python
# replace(old, new) returns a new string with the replacements made.
'Hi, my name is madhu'.replace('madhu', 'dadi')
```

```python
# No match, so the original string is returned unchanged.
'Hi, my name is madhu'.replace('madhukumar', 'dadi')
```

##### Strip

```python
# strip() removes leading and trailing whitespace.
'madhu              '.strip()
```

```python
# Whitespace is trimmed from both ends.
'           madhu             '.strip()
```
