##### Creating Strings

```python
# Strings can be created with single quotes.
# Output: hello -> the string is printed as-is.
s = 'hello'
print(s)
```

```python
# Double quotes work the same way.
# Output: hello -> double quotes give the same result as single quotes.
s = "hello"
print(s)
```

```python
# Triple-quoted strings can span multiple lines.
# for multiline strings
# Output: hello -> triple quotes create the same string; they only matter for text spanning several lines.
s = """hello"""
print(s)
```

```python
# An apostrophe inside single quotes ends the string early -> SyntaxError.
# Output: SyntaxError - the second apostrophe closes the string early, leaving stray text after it.
'it's raining outside'
```

```python
# Double quotes avoid the need to escape the apostrophe.
# Output: "it's raining outside" -> double quotes let an apostrophe through without escaping.
"it's raining outside"
```

```python
# The str() constructor also builds a string.
# Output: hello -> str('hello') simply returns the same string.
s = str('hello')
print(s)
```

##### Accessing Substrings from a String

```python
# Indexing starts at 0, so s[0] is the first character.
# Indexing
# Output: h -> index 0 is the first character.

s = 'hello world'
print(s[0])
```

```python
# s[4] is the fifth character.
# Output: o -> index 4 is the fifth character.
print(s[4])
```

```python
# An index outside the string raises an IndexError.
# Output: IndexError - 'hello world' has only 11 characters (indices 0-10), so 41 is out of range.
print(s[41])
```

```python
# A negative index counts from the end; -1 is the last character.
# Negative Indexing
# Output: d -> -1 counts from the end of the string.
print(s[-1])
```

```python
# -2 is the second character from the end.
# Output: l -> -2 is the second character from the end.
print(s[-2])
```

```python
# Slicing s[start:stop]; the stop index is excluded.
# Slicing
# Output: hello -> indices 0 up to (not including) 5.

print(s[0:5])
```

```python
# Characters from index 6 up to (not including) index 11.
# Output: world -> indices 6 up to (not including) 11.
print(s[6:11])
```

```python
# Stops before index 10, so the final d is left out.
# Output: worl -> slicing stops before index 10, so the final 'd' is left out.
print(s[6:10])
```

```python
# Leaving out stop slices through to the end.
# Output: llo world -> from index 2 through to the end.
print(s[2:])
```

```python
# From index 6 to the end.
# Output: world -> from index 6 through to the end.
print(s[6:])
```

```python
# A copy of the whole string.
# Output: hello world -> the whole string.
print(s[:])
```

```python
# The third slice value is the step (every 2nd character).
# Output: hlo -> every 2nd character of 'hello' (h, l, o).
print(s[0:5:2])
```

```python
# A step of 3.
# Output: hlw -> every 3rd character up to index 7 (h, l, w).
print(s[0:8:3])
```

```python
# A step of 1 is the default.
# Output: hello -> a step of 1 is the default.
print(s[0:5:1])
```

```python
# A negative step with start before stop returns an empty string.
# Output: an empty string -> with a negative step the start (0) is already past the stop (5), so nothing is selected.
print(s[0:5:-1])
```

```python
# A negative step walks backwards; stop (0) is excluded.
# Output: ' olle' -> walking backwards from index 5 down to (but not including) 0.
print(s[5:0:-1])
```

```python
# Backwards with a step of 2.
# Output: ' le' -> the same backwards walk with a step of 2.
print(s[5:0:-2])
```

```python
# Idiom to reverse a string.
# Output: dlrow olleh -> a step of -1 reverses the whole string.
print(s[::-1])
```

```python
# Slice using negative start and stop.
# Output: worl -> from the 5th-from-last up to (not including) the last character.
print(s[-5:-1])
```

```python
# The last five characters.
# Output: world -> the last five characters.
print(s[-5:])
```

```python
# The last five characters, reversed.
# Output: dlrow -> the last five characters in reverse.
print(s[-1:-6:-1])
```

##### Editing and Deleting in Strings

```python
# Set up s again.
# Output: nothing -> the cell only (re)creates the variable s for the examples that follow.
s = 'hello world'
```

```python
# Strings are immutable, so item assignment raises a TypeError.
# Output: TypeError - strings are immutable, so one character cannot be replaced; build a new string instead.
s[0] = 'H'  # Python strings are immutable
```

```python
# Re-create s after the deletion example.
# Output: nothing -> the cell only (re)creates the variable s after the deletion example.
s = 'hello world'
```

```python
# del removes the name; using it afterwards raises a NameError.
# Output: NameError - 'del' removes the name s, so it no longer exists.
del s
print(s)
```

```python
# Slices cannot be deleted -> TypeError.
# Output: TypeError - 'del' cannot remove a slice of a string (strings are immutable).
s = 'hello world'
del s[-1:-5:2]
print(s)
```

##### Operations on Strings
* Arithmetic Operations
* Relational Operations
* Logical Operations
* Loops on Strings

```python
# The + operator concatenates strings.
# Output: delhimumbai -> '+' joins the two strings with nothing between them.
print('delhi' + 'mumbai')
```

```python
# Concatenate with a space in between.
# Output: delhi mumbai -> a space string is concatenated in the middle.
print('delhi' + ' ' + 'mumbai')
```

```python
# The * operator repeats a string.
# Output: delhi repeated five times -> '*' repeats the string.
print('delhi'*5)
```

```python
# Repeating a character to draw a separator line.
# Output: exactly 50 asterisks -> '*' repeated 50 times, handy for a separator line.
print('*'*50)
```

```python
# Equality is case sensitive.
# Output: False -> the two strings are different.
'delhi' == 'mumbai'
```

```python
# Not-equal comparison.
# Output: False -> the strings are equal, so '!=' is False.
'delhi' != 'delhi'
```

```python
# A difference in case makes the strings unequal.
# Output: False -> comparison is case sensitive.
'delhi' == 'Delhi'
```

```python
# Strings compare lexicographically, character by character.
# Output: False -> compared letter by letter, 'm' (109) is smaller than 'p' (112).
'mumbai' > 'pune'   # lexiographically comparison
```

```python
# Uppercase letters have smaller codes than lowercase, so P < p.
# Output: False -> uppercase letters have smaller codes than lowercase, so 'P' (80) < 'p' (112).
'Pune' > 'pune'  # https://www.littlebeartools.com/asciicode/
```

```python
# 'and' returns the first falsy operand, otherwise the last operand.
# Output: 'world' -> 'and' returns the last operand when all operands are truthy.
'hello' and 'world'
```

```python
# 'or' returns the first truthy operand.
# Output: 'hello' -> 'or' returns the first truthy operand.
'hello' or 'world'
```

```python
# An empty string is falsy.
# Output: '' -> 'and' stops at the falsy empty string and returns it.
'' and 'world'
```

```python
# So 'or' returns the non-empty string.
# Output: 'world' -> 'or' skips the falsy empty string and returns the next one.
'' or 'world'
```

```python
# 'not' negates the truthiness of a value.
# Output: False -> a non-empty string is truthy, so 'not' gives False.
not 'world'
```

```python
# An empty string is falsy, so not '' is True.
# Output: True -> an empty string is falsy, so 'not' gives True.
not ''
```

```python
# A string can be iterated character by character.
# Output: h, e, l, l, o on separate lines -> the loop reads one character per pass.
for i in 'hello':
    print(i)
```

```python
# The body runs once for each character in the string.
# Output: 'pune' five times -> the body runs once per character even though it ignores i.
for i in 'delhi':
    print('pune')
```

```python
# Membership is case sensitive.
# Output: False -> membership is case sensitive ('d' is present, 'D' is not).
'D' in 'delhi'
```

```python
# The empty string is considered to be in every string.
# Output: True -> the empty string is treated as present in every string.
'' in 'delhi'
```

```python
# There is no space in the string, so this is False.
# Output: False -> there is no space inside 'delhi'.
' ' in 'delhi'
```

##### Format

```python
# str.format() fills the {} placeholders with the arguments in order.
# Output: 'Hi, my name is madhu and I am a male' -> the two {} are filled with the arguments in order.
name = 'madhu'
gender = 'male'

'Hi, my name is {} and I am a {}'.format(name, gender)
```

```python
# Placeholders are filled in the order the arguments are given.
# Output: 'Hi, my name is male and I am a madhu' -> the placeholders follow the argument order, so the values are swapped.
'Hi, my name is {} and I am a {}'.format(gender, name)
```

```python
# Numbered placeholders refer to arguments by position.
# Output: the correct sentence -> the numbers pick the arguments explicitly, restoring the order.
'Hi, my name is {0} and I am a {1}'.format(name, gender)
```

```python
# The numbering can reorder the arguments.
# Output: 'Hi, my name is madhu and I am a male' -> using index 1 first picks the second argument.
'Hi, my name is {1} and I am a {0}'.format(gender, name)
```

```python
# The same index may be reused in the template.
# Output: 'Hi, my name is madhu and I am a madhu' -> the same index can be reused.
'Hi, my name is {1} and I am a {1}'.format(gender, name)
```

##### Split / Join

```python
# split() with no argument splits on whitespace.
# Output: the sentence split into words at the spaces -> split() with no argument breaks on whitespace.
'Hi, my name is madhu'.split()
```

```python
# split(sep) splits on the given separator.
# Output: ['Hi', ' my name is madhu'] -> the string is cut only at the comma, so the rest stays together.
'Hi, my name is madhu'.split(',')
```

```python
# Join with a hyphen separator.
# Output: 'Hi,-my-name-is-madhu' -> the separator between items is a hyphen.
"-".join(['Hi,', 'my', 'name', 'is', 'madhu'])
```

### Example Programs

```python
# Find the length of a given string without using the len() function
# Why/how: len() is not used - the loop visits every character and adds 1, so the counter equals the number of characters.
# Output: 'length of string is 5' -> a 5-character input counts up to 5.

s = input('Enter the string')

counter = 0

for i in s:
    counter += 1

print('length of string is', counter)
```

```python
# Extract username from a given email
# Eg if the email is madhu@gmail.com
# then username should be madhu
# Why/how: index('@') finds the position of '@' and the slice s[0:pos] keeps everything before it.
# Output: 'madhu' -> the part of the email before the '@'.

s = input('Enter the email')

pos = s.index('@')
print(s[0:pos])
```

```python
# Count the frequency of a particular character in a provided string.
# Eg 'hello how are you' is the string, the frequency of h in this string is 2.
# Why/how: the loop compares every character with the search term and increments the counter on a match.
# Output: 'frequency 2' -> the searched character occurred twice.

s = input('Enter a string')
term = input('What would you like to search for?')

counter = 0
for i in s:
    if i == term:
        counter += 1

print('frequency', counter)
```

```python
# write a program which can remove a particular character from a string
# Why/how: characters that equal the term are skipped, the rest are appended to result.
# Output: 'i ow are you' -> every 'h' was removed from the string.

s = input('Enter a string')
term = input('What would you like to remove?')

result = ''

for i in s:
    if i != term:
        result = result + i

print(result)
```

```python
# Write a program that can check whether a given string is palindrome or not.
# Why/how: each character is compared with its mirror from the end; a mismatch sets flag to False.
# Output: 'Not a Palidrome' -> at least one mirrored pair of characters did not match.

# abba
# malayalam

s = input('Enter a string')
flag = True

for i in range(0,len(s)//2):
    if s[i] != s[len(s) - i -1]:
        flag = False
        break

if flag:
    print('Palindrome')
else:
    print('Not a Palidrome')
```

```python
# Write a program to count the number of words in a string without split()
# Why/how: characters are gathered into temp until a space is met, then the word is appended to the list.
# Output: ['hi', 'how', 'are', 'you'] -> one list item per word, without using split().

s = input('Enter a string')
L = []
temp = ''

for i in s:
    if i != ' ':
        temp = temp + i
    else:
        L.append(temp)
        temp = ''

L.append(temp)

print(L)
```

```python
# Write a python program to convert a string to title case without using the title()
# Why/how: each word is rebuilt as first-letter-upper plus the rest lower-cased.
# Output: the capitalised list, then the words joined by spaces -> title case without title().

s = input('enter the string')
L = []

for i in s.split():
    L.append(i[0].upper() + i[1:].lower())

print(L)
print(" ".join(L))
```

```python
# Write a program that can convert an integer to string
# Why/how: each digit is taken with %10 and looked up in the string '0123456789', so the result is a str.
# Output: 1456 and <class 'str'> -> the number is rebuilt as text, which type() confirms.

number = int(input('enter the number'))

digits = '0123456789'
result = ''

while number != 0:
    result = digits[number % 10] + result
    number = number // 10

print(result)
print(type(result))
```
