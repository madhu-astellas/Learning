### 1. Python Output

```python
# print() writes text or values to the console (standard output).
# Python is a case sensitive Language

print("Hello World")
```

```python
# Strings may be written with single or double quotes - both create a str.
print('Madhu Dadi')
```

```python
# Quotes are missing, so Python reads Madhu and Dadi as names -> SyntaxError.
print(Madhu Dadi)
```

```python
# 'Madhu' was never defined -> NameError: name 'Madhu' is not defined.
print(Madhu)
```

```python
# Integer (int) literal - a whole number with no decimal point.
print(1)
```

```python
# Float (float) literal - a number with a decimal point.
print(1.3)
```

```python
# Boolean (bool) literal - only True or False (capital first letter).
print(True)
```

```python
# The other boolean value, False.
print(False)
```

```python
# print() takes several arguments separated by commas (default separator is a space).
print('Hello',1,4.5,True)
```

```python
# sep='/' sets the separator printed between the arguments.
print('Hello',1,4.5,True, sep='/')
```

```python
# sep can be any string; here a hyphen is used.
print('Hello',1,4.5,True, sep="-")
```

```python
# Each print() ends with a newline by default, so output appears on separate lines.
print("Madhu")
print("Dadi")
```

```python
# sep only matters within a single print() call with multiple arguments.
print("Madhu", sep="-")
print("Dadi")
```

```python
# end controls what is printed after the call; its default value is a newline.
print("Madhu", end="\n")
print("Dadi")
```

```python
# end=' ' replaces the newline with a space so the next print stays on the same line.
print("Madhu", end=" ")
print("Dadi")
```

```python
# end='-' joins the two prints with a hyphen on one line.
print("Madhu", end='-')
print("Dadi")
```

### 2. Data Types

### 3. Variables

```python
# Static typing: the type must be declared up front, as in C/C++.
# Static Vs Dynamic Typing

# Static Typing - Will tell the data type

# In C/C++
# int name = 'Madhu'
```

```python
# Dynamic typing: no declaration is needed - Python infers the type at run time.
# Dynamic Typing - Do not tell the data type

name = "Madhu"
print(name)

a = 5
b = 6

print(a + b)
```

```python
# Dynamic binding: the same variable can be re-bound to a different type later.
# Dynamic Binding - Same variable can have data of different data types

a = 5
print(a)

a = "Madhu"
print(a)
```

```python
# Static binding: in C/C++/Java a variable keeps the type it was declared with.
# Static Binding - In C, C++, Java

# int a = 5
```

```python
# Plain assignment of three variables on separate lines.
# Stylish Declaration Techniques

a = 1
b = 2
c = 3
print(a,b,c)
```

```python
# Tuple unpacking: assign several variables in one statement.
a,b,c = 1,2,3
print(a,b,c)
```

```python
# Chained assignment: a, b and c all receive the value 5.
a=b=c=5
print(a,b,c)
```

### 4. Keywords & Identifiers

#### Keywords


#### Identifier
##### You can't start with a digit
##### You can use special chars "_" only
##### Identifiers can not be keywords

```python
# An identifier may contain letters and digits but must not start with a digit.
name1 = "Madhu"
print(name1)
```

```python
# Starting a name with a digit is illegal -> SyntaxError.
1name = "Madhu"
```

```python
# Underscores are allowed inside identifiers (snake_case).
first_name = "Madhu"
print(first_name)
```

```python
# A hyphen is not allowed in a name - it is read as subtraction -> SyntaxError.
first-name = "Madhu"
```

```python
# A lone underscore is a legal (but unusual) variable name.
_ = "Madhu"
print(_)
```

### 5. User Input

```python
# input() reads a line typed by the user (no prompt here).
# Static vs Dynamic

input()
```

```python
# The argument to input() is shown as the prompt.
input('What is your name?')
```

```python
# input() always returns a string, so '+' concatenates instead of adding numbers.
# Take input from users and store them in a variable
fnum = input('Enter first number')
snum = input('Enter second number')
print(fnum, snum)

# add 2 variables
result = fnum + snum

# print result
print(result)

print(type(fnum), type(snum))
```

### 6. Type Conversion

```python
# Implicit conversion: int + float automatically becomes float.
# Implicit vs Explicit

# Implicit
print(5 + 5.6)
print(type(5), type(5.6))
```

```python
# int + str is not allowed -> TypeError; an explicit conversion is needed.
# Explicit
print(4 + '4')
```

```python
# int(4.5) truncates the decimal part (it does not round) -> 4.
# str -> int
int(4.5)
```

```python
# A complex number cannot be converted to int -> TypeError.
int(4+5j)
```

```python
# str() converts a value to a string.
# int -> str
str(5)
```

```python
# float() converts an int to a float.
# float
float(4)
```

```python
# Casting the inputs to int so that '+' performs numeric addition.
# Take input from users and store them in a variable
fnum = input('Enter first number')
snum = input('Enter second number')
print(fnum, snum)

# add 2 variables
result = int(fnum) + int(snum)

# print result
print(result)

print(type(fnum), type(snum))
```

```python
# Same result, but the cast happens right at the input() call.
# Take input from users and store them in a variable
fnum = int(input('Enter first number'))
snum = int(input('Enter second number'))
print(fnum, snum)

# add 2 variables
result = fnum + snum

# print result
print(result)

print(type(fnum), type(snum))
```

### 7. Literals

```python
# Number literals in other bases: binary 0b, octal 0o, hexadecimal 0x.
a = 0b1010 # Binary Literals
b = 100 # Decimal Literal
c = 0o310 # Octal Literal
d = 0x12c # Hexadecimal Literal
```

```python
# Every base prints the same value in decimal.
print(a, b, c, d)
```

```python
# Float literals, including scientific notation (1.5e2 = 150.0).
# Float Literal

float_1 = 10.5
float_2 = 1.5e2
float_3 = 1.5e-3

print(float_1, float_2, float_3)
```

```python
# Complex literal; .real and .imag expose the real and imaginary parts.
# Complex Literal

x = 3.14j

print(x, x.imag, x.real)
```

```python
# String literal variants: quotes, triple-quoted, unicode (u) and raw (r).
string = 'This is Python'
strings = "This is Python"
char = "C"
multiline_str = """This is a multiline string with more than one line code."""
unicode = u"\U0001f600\U0001F606\U0001F923"
raw_str = r"raw \n string"

print(string)
print(strings)
print(char)
print(multiline_str)
print(unicode)
print(raw_str)
```

```python
# True acts as 1 and False acts as 0 in arithmetic.
a = True + 4
b = False + 10

print("a:", a)
print("b:", b)
```

```python
# None is the special object that represents 'no value'.
a = None
print(a)
```

```python
# None cannot take part in arithmetic -> TypeError.
a = None + 1
print(a)
```
